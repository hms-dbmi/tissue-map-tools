```python
from pathlib import Path

from tissue_map_tools.view import view_precomputed_in_vitessce

##
# Path to a precomputed dataset produced by one of the other examples
# (e.g. examples/invasive/ome_tiff_to_sharded_meshes.py), containing a
# segmentation + meshes, and optionally point annotations alongside it.
precomputed_path = Path("/Users/tabassumkakar/Development/tissue-map-tools/data/invasive_precomputed")

if not precomputed_path.exists():
    raise FileNotFoundError(
        f"{precomputed_path} does not exist. Run the invasive mesh-generation "
        "example first, or point precomputed_path at your own precomputed data."
    )

##
# show_meshes=True adds the segmentation + meshes, wired up via the
# segmentationLayer/segmentationChannel coordination pattern.
# show_annotations=True looks for and adds any point annotation layers found
# alongside the precomputed data.
# host_local_data=True starts a local CloudVolume server (in a background
# thread) to serve the precomputed directory over HTTP, since it's a plain
# local path rather than an already-reachable URL.
widget = view_precomputed_in_vitessce(
    data_path=str(precomputed_path),
    show_meshes=True,
    show_annotations=True,
    port=10001,
    host="localhost",
    name="INV",
    host_local_data=True,
    # mesh_ids=['2', '5', '100', '199']
    segments =[612, 3351, 4328, 6531, 8446]
)

##
widget
```
